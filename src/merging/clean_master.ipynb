```python
import pandas as pd
df = pd.read_csv('../../data/merged/master_dataset.csv')
```

```python
df.columns
```

```python
# dropping columns we don't need for analysis
columns_to_drop = [
      'master_id',
      'title_normalized',
      'score_tmdb_numbers',
       'score_master_rotten', 'tmdb_tmdb_id', 'tmdb_title',
       'tmdb_original_title',
       'numbers_title',
       'rotten_title',
       
]
```

```python
df.drop(columns=columns_to_drop, inplace=True)
```

```python
df.columns
```

```python
# removing the tmdb_ , numbers_ and rotten_ prefixes from the column names
df.columns = df.columns.str.replace(r'^(tmdb_|numbers_|rotten_)', '', regex=True)
```

```python
df.columns
```

```python
# save the df as df_for_analysis.csv
df.to_csv('../../data/merged/df_for_analysis.csv', index=False)
```
