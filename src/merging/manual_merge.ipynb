```python
import pandas as pd
import os
```

```python
numbers_rotten_100_df = pd.read_csv("../../data/merged/score_100/numbers_rotten_100.csv")
numbers_rotten_scores_df = pd.read_csv("../../data/merged/similarities/numbers_rotten_scores.csv")

all_three_100_df = pd.read_csv("../../data/merged/score_100/all_three_100.csv")
all_three_scores_df = pd.read_csv("../../data/merged/similarities/all_three_scores.csv")

tmdb_numbers_100_df = pd.read_csv("../../data/merged/score_100/tmdb_numbers_100.csv")
tmdb_numbers_scores_df = pd.read_csv("../../data/merged/similarities/tmdb_numbers_scores.csv")

tmdb_rotten_100_df = pd.read_csv("../../data/merged/score_100/tmdb_rotten_100.csv")
tmdb_rotten_scores_df = pd.read_csv("../../data/merged/similarities/tmdb_rotten_scores.csv")
```

## Merge numbers and rotten

```python

numbers_rotten_scores_df[numbers_rotten_scores_df["similarity"]<100].head(20)
```

```python
rotten_df = pd.read_csv("../../data/processed/rotten_tomatoes_cleaned.csv")
numbers_df = pd.read_csv("../../data/processed/the_numbers_worldwide_2026_cleaned.csv")
tmdb_df = pd.read_csv("../../data/processed/tmdb_clean.csv")


# once upon a timein hollywood
merged_1 = pd.concat([
    rotten_df.iloc[2],
    numbers_df.iloc[22]
]).to_frame().T
# billie eilish hit me hard and softthe tour
merged_2 = pd.concat([
    rotten_df.iloc[33509],
    numbers_df.iloc[676]
]).to_frame().T


manual_merge_rotten_numbers_df = pd.concat([merged_1, merged_2], ignore_index=True)
```

```python
manual_merge_rotten_numbers_df = manual_merge_rotten_numbers_df[
    manual_merge_rotten_numbers_df.columns.intersection(numbers_rotten_100_df.columns)
]
```

```python
manual_merge_rotten_numbers_df = manual_merge_rotten_numbers_df.loc[:, ~manual_merge_rotten_numbers_df.columns.duplicated()]
```

```python
final_manual_merge_rotten_numbers_df = pd.concat(
    [numbers_rotten_100_df, manual_merge_rotten_numbers_df],
    axis=0,
    ignore_index=True

)
```

## merge tmdb scores

```python

tmdb_numbers_scores_df[tmdb_numbers_scores_df["similarity"]<100][19:30]
```

```python
tmdb_indexes = [9934,39869,59691,2,69806,29822,6579,39809,19866,30383,59688]
numbers_indexed = [100,321,510,22,635,225,32,310,758,274,512]
merged_rows = []

for tmdb_idx, numbers_idx in zip(tmdb_indexes, numbers_indexed):
    merged_row = pd.concat([
        tmdb_df.iloc[tmdb_idx],
        numbers_df.iloc[numbers_idx]
    ]).to_frame().T
    
    merged_rows.append(merged_row)

merged_manual_numbers_tmdb_df = pd.concat(merged_rows, ignore_index=True)
```

```python
merged_manual_numbers_tmdb_df 
```

```python
# remove duplicated columns
merged_manual_numbers_tmdb_df = merged_manual_numbers_tmdb_df.loc[:, ~merged_manual_numbers_tmdb_df.columns.duplicated()]
```

```python
merged_manual_numbers_tmdb_df  = merged_manual_numbers_tmdb_df [
    merged_manual_numbers_tmdb_df.columns.intersection(tmdb_numbers_100_df.columns)
]
final_merged_manual_numbers_tmdb_df = pd.concat(
    [tmdb_numbers_100_df, merged_manual_numbers_tmdb_df],
    axis=0,
)
```

## merge rotten and tmdb

```python
tmdb_rotten_scores_df[tmdb_rotten_scores_df["similarity"]<100]
```

```python
tmdb_indexes = [1193,5503,5360]
rotten_indexed = [877,2834,3038]
merged_rows = []

for tmdb_idx,rotten_idx in zip(tmdb_indexes, rotten_indexed):
    merged_row = pd.concat([
        tmdb_df.iloc[tmdb_idx],
        rotten_df.iloc[rotten_idx]
    ]).to_frame().T
    
    merged_rows.append(merged_row)

merged_manual_rotten_tmdb_df = pd.concat(merged_rows, ignore_index=True)
```

```python
merged_manual_rotten_tmdb_df = merged_manual_rotten_tmdb_df.loc[:, ~merged_manual_rotten_tmdb_df.columns.duplicated()]
```

```python
merged_manual_rotten_tmdb_df  = merged_manual_rotten_tmdb_df[
    merged_manual_rotten_tmdb_df.columns.intersection(merged_manual_rotten_tmdb_df.columns)
]
final_merged_manual_rotten_tmdb_df = pd.concat(
    [tmdb_numbers_100_df, merged_manual_rotten_tmdb_df],
    axis=0,
)
```
