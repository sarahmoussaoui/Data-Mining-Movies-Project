```python
import re
import unicodedata

def normalize_title(title):
    title = str(title).lower()

    # Remove accents
    title = unicodedata.normalize('NFKD', title)
    title = ''.join(c for c in title if not unicodedata.combining(c))

    # Remove punctuation
    title = re.sub(r'[^a-z0-9\s]', '', title)

    # Remove extra spaces
    title = re.sub(r'\s+', ' ', title).strip()

    return title
```

```python
tmdb_df['title_normalized'] = tmdb_df['title'].apply(normalize_title)
rotten_df['title_normalized'] = rotten_df['title'].apply(normalize_title)
numbers_df['title_normalized'] = numbers_df['title'].apply(normalize_title)
```

```python
from rapidfuzz import process
from rapidfuzz.distance import Levenshtein

numbers_titles = numbers_df['title_normalized'].tolist()

matches = []

for _, tmdb_row in tmdb_df.iterrows():

    tmdb_title = tmdb_row['title_normalized']

    match = process.extractOne(
        tmdb_title,
        numbers_titles,
        scorer=Levenshtein.normalized_similarity
    )

    matches.append({
        'tmdb_title': tmdb_row['title'],
        'tmdb_normalized': tmdb_title,
        'numbers_title': match[0],
        'similarity': match[1] * 100
    })

tmdb_numbers_fuzzy = pd.DataFrame(matches)

display(
    tmdb_numbers_fuzzy.sort_values(
        'similarity',
        ascending=False
    )
)
```

```python
def normalize_title(title):
    title = str(title).lower()

    # Remove accents
    title = unicodedata.normalize('NFKD', title)
    title = ''.join(c for c in title if not unicodedata.combining(c))

    # Remove punctuation
    title = re.sub(r'[^a-z\s]', '', title)

    # Remove extra spaces
    title = re.sub(r'\s+', ' ', title).strip()

    return title

tmdb_df['title_normalized'] = tmdb_df['title'].apply(normalize_title)
rotten_df['title_normalized'] = rotten_df['title'].apply(normalize_title)
numbers_df['title_normalized'] = numbers_df['title'].apply(normalize_title)
```

now what if we take the year into consideration, how many exact macthes will we have ? 

```python
from rapidfuzz import process
from rapidfuzz.distance import Levenshtein

matches = []

for _, tmdb_row in tmdb_df.iterrows():

    tmdb_title = tmdb_row['title_normalized']
    tmdb_year = tmdb_row['year']

    # Keep only Numbers movies with the same year
    numbers_same_year = numbers_df[
        numbers_df['year'] == tmdb_year
    ]

    # If there are no movies with this year, skip
    if numbers_same_year.empty:
        continue

    # Titles from Numbers for this year only
    numbers_titles = numbers_same_year['title_normalized'].tolist()

    # Fuzzy match
    match = process.extractOne(
        tmdb_title,
        numbers_titles,
        scorer=Levenshtein.normalized_similarity
    )

    # Retrieve the corresponding Numbers row
    numbers_row = numbers_same_year[
        numbers_same_year['title_normalized'] == match[0]
    ].iloc[0]

    matches.append({
        'tmdb_title': tmdb_row['title'],
        'tmdb_normalized': tmdb_title,
        'tmdb_year': tmdb_year,
        'numbers_title': numbers_row['title'],
        'numbers_normalized': numbers_row['title_normalized'],
        'numbers_year': numbers_row['year'],
        'similarity': match[1] * 100
    })

tmdb_numbers_fuzzy = pd.DataFrame(matches)

display(
    tmdb_numbers_fuzzy.sort_values(
        'similarity',
        ascending=False
    )
)
```

```python
number_100_matches = (tmdb_numbers_fuzzy['similarity'] == 100).sum()

print("Number of 100% matches:", number_100_matches)
```

trying the difflib library

```python
import difflib

matches = []

for _, tmdb_row in tmdb_df.iterrows():

    tmdb_title = tmdb_row['title_normalized']
    tmdb_year = tmdb_row['year']

    # Keep only Numbers movies with the same year
    numbers_same_year = numbers_df[
        numbers_df['year'] == tmdb_year
    ]

    # If there are no movies with this year, skip
    if numbers_same_year.empty:
        continue

    best_match = None
    best_score = -1

    # Compare with all Numbers titles from the same year
    for _, numbers_row in numbers_same_year.iterrows():

        numbers_title = numbers_row['title_normalized']

        score = difflib.SequenceMatcher(
            None,
            tmdb_title,
            numbers_title
        ).ratio()

        if score > best_score:
            best_score = score
            best_match = numbers_row

    matches.append({
        'tmdb_title': tmdb_row['title'],
        'tmdb_normalized': tmdb_title,
        'tmdb_year': tmdb_year,
        'numbers_title': best_match['title'],
        'numbers_normalized': best_match['title_normalized'],
        'numbers_year': best_match['year'],
        'similarity': best_score * 100
    })

tmdb_numbers_difflib = pd.DataFrame(matches)

display(
    tmdb_numbers_difflib.sort_values(
        'similarity',
        ascending=False
    )
)
```

```python
number_100_matches = (
    tmdb_numbers_difflib['similarity'] == 100
).sum()

print("Number of 100% matches:", number_100_matches)
```
